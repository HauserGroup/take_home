1. Import the python libraries: pandas, matplotlib, numpy and scipy curve_fit

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit
```

2. Import the data using pandas

```python
df = pd.read_csv('PD_tutorial_data.csv')
print(df)

concentration = np.array(df['Isoprenaline concentration (M)'])

effect = np.array(df['Increase in heart rate (beats/min)'])
```

3. Define the Hill equation. ACTION: check that this is the correct equation.

```python
def Hill_equation(C, E_max, EC_50, g): #Hill Langmuir equation
    E = E_max*(C**g)/(EC_50**g+C**g)
    return E
```

4. Plot the raw data. 
ACTION: To get a log x-axis, you should insert the line:
plt.xscale("log")


```python

plt.figure(figsize=(8, 5))

plt.scatter(concentration, effect, label='Observed data', color='black')

plt.xlabel('Isoprenaline concentration (M)')
plt.ylabel('Increase in heart rate (beats/min)')



plt.legend()
plt.tight_layout()
plt.show()
```

5. You want to use the Hill-Langmuir equation to describe the relationship between concentration of drug and biological effect. Three parameters are needed to do this: Emax, EC50 and gamma (Hill slope). I have set them to 10, 1e-5 and 5, respectively.

```python
E_max_guess =  10  #this is just a guess
EC_50_guess =   1e-5   #this is just a guess
g_guess = 5  #this is just a guess

E_guess = [Hill_equation(i, E_max_guess, EC_50_guess, g_guess) for i in concentration] # This will take the concentration values from the table and calculate what the biological effect would be for your guess parameters


print(E_guess) #prints the E values for each concentration, using your guess values of E_max, EC_50 and g
```

6. Plot the values of E that guesses would produce. QUESTION: Were my guesses good? ACTION: Go back and change the guess values to try and get a better match with the observed data (go back to 5. and change the numbers, run cells 5. and 6. again).

```python
#plot with guess
plt.figure(figsize=(8,5))


plt.scatter(concentration, effect,  label='Observed data', color='black')
plt.plot(concentration, E_guess, label='Guess', color='green')

plt.xlabel('Isoprenaline concentration (M)')
plt.ylabel('Increase in heart rate (beats/min)')

plt.xscale("log")
plt.legend()
plt.tight_layout()
plt.show()
```

7. Now you will fit the data. The fitting procedure will start the guess, calculate how well it matches the observed data. Then it will change the values, calculate it it matches better. It repeats this procedure until the values stop improving, and it will output the best fitting values. The "fit" values for E_max, EC_50 and g.

```python
popt, pcov = curve_fit(Hill_equation, concentration, effect, p0=(E_max_guess, EC_50_guess,g_guess))

E_max_fit, EC_50_fit, g_fit  = popt
E_max_fit_err, EC_50_fit_err, g_fit_err = np.sqrt(np.diag(pcov))

print('E0_fit = '+str(E_max_fit)+' ± '+ str(E_max_fit_err))
print('EC_50_fit = '+str(EC_50_fit)+' ± '+ str(EC_50_fit_err))
print('g_fit = '+str(g_fit)+' ± '+ str(g_fit_err))

E_fit = [Hill_equation(i, E_max_fit, EC_50_fit, g_fit) for i in concentration]
```

9. Now you will plot what effect would look like using the fit values of E_max, EC_50 and g.

```python
#plot with guess
plt.figure(figsize=(8,5))


plt.scatter(concentration, effect,  label='Observed data', color='black')
plt.plot(concentration, E_fit, label='Fit', color='red')

plt.xlabel('Isoprenaline concentration (M)')
plt.ylabel('Increase in heart rate (beats/min)')

plt.xscale("log")
plt.legend()
plt.tight_layout()
plt.show()
```
